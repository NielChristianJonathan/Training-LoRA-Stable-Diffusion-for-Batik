```python
import os
import torch
from diffusers import StableDiffusionPipeline

assert torch.cuda.is_available(), "GPU is required to run this notebook."

OUTPUT_DIR = "../../results/baseline_no_lora"
os.makedirs(OUTPUT_DIR, exist_ok=True)

print("✅ Environment ready")
print("📁 Output directory:", OUTPUT_DIR)
```

```python
BASE_MODEL = "runwayml/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    BASE_MODEL,
    torch_dtype=torch.float16
).to("cuda")

pipe.enable_attention_slicing()

print("✅ Base model loaded (no LoRA)")
```

```python
prompt = "traditional batik megamendung pattern"
generator = torch.Generator("cuda").manual_seed(42)

image = pipe(
    prompt,
    num_inference_steps=30,
    guidance_scale=7.5,
    generator=generator
).images[0]

image.save(f"{OUTPUT_DIR}/baseline_megamendung.png")
image
```

## Baseline Result (No LoRA)

This image is generated using Stable Diffusion v1.5 without any fine-tuning.
The model does not strongly capture the distinctive cloud-like structure of
Megamendung batik, serving as a baseline reference for comparison with LoRA.
